```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# load csv
csv_path = Path("../results/feasibility/results.csv")
df = pd.read_csv(csv_path)
df.head()

# plot bfs
workload = "bfs"
data = df[df["workload"] == workload]
data

fig, axes = plt.subplots(2, 1, figsize=(10, 8))

# IPC
axes[0].bar(data["scheduler"], data["ipc"])
axes[0].set_title(f"{workload}: IPC by Scheduler")
axes[0].set_ylabel("IPC")
axes[0].set_xlabel("Scheduler")

# Instruction count
axes[1].bar(data["scheduler"], data["instructions"])
axes[1].set_title(f"{workload}: Instruction Count by Scheduler")
axes[1].set_ylabel("Instructions")
axes[1].set_xlabel("Scheduler")

plt.tight_layout()
plt.show()
```
